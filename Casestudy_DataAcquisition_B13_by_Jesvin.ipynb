```python
# Q1. Read the Population dataset

import pandas as pd

population = pd.read_csv("/content/drive/MyDrive/DSAB13/casestudy/materials/population.csv")

print(population.head())
print(population.shape)
print(population.dtypes)
```

```python
# Q2. Read the GDP dataset

gdp = pd.read_excel("/content/drive/MyDrive/DSAB13/casestudy/materials/gdp.xlsx")

print(gdp.head())
print(gdp.columns)
print(gdp.dtypes)
```

```python
# Q3. Read the Internet Users dataset

internet = pd.read_json("/content/drive/MyDrive/DSAB13/casestudy/materials/internet_users.json")

print(internet.head())
```

```python
# Q4. Read the Literacy Rate dataset from XML

literacy = pd.read_xml("/content/drive/MyDrive/DSAB13/casestudy/materials/literacy_rate.xml")

print(literacy.head())
print("Shape:", literacy.shape)
```

```python
# Q5. Explore each dataset

datasets = {
    "Population": population,
    "GDP": gdp,
    "Internet": internet,
    "Literacy": literacy
}

for name, data in datasets.items():

    print("\nDataset:", name)

    print("Rows:", data.shape[0])
    print("Columns:", data.shape[1])

    print("Column Names:")
    print(data.columns.tolist())

    print("Data Types:")
    print(data.dtypes)

    print("Missing Values:")
    print(data.isnull().sum())

    print("Duplicate Records:",
          data.duplicated().sum())
```

```python
# ============================================================
# PART B - DATA UNDERSTANDING
# ============================================================


# ------------------------------------------------------------
# Q5. Explore each dataset
# Find:
# 1. Number of rows
# 2. Number of columns
# 3. Column names
# 4. Data types
# 5. Missing values
# 6. Duplicate records
# ------------------------------------------------------------

datasets = {
    "Population": population,
    "GDP": gdp,
    "Internet Users": internet,
    "Literacy Rate": literacy
}

for name, data in datasets.items():

    print("\n================================")
    print("Dataset:", name)
    print("================================")

    print("Number of Rows:", data.shape[0])

    print("Number of Columns:", data.shape[1])

    print("Column Names:")
    print(data.columns.tolist())

    print("Data Types:")
    print(data.dtypes)

    print("Missing Values:")
    print(data.isnull().sum())

    print("Duplicate Records:")
    print(data.duplicated().sum())
```

```python
# ============================================================
# PART C - DATA TRANSFORMATION
# ============================================================


# ------------------------------------------------------------
# Q6. Rename columns wherever necessary
# Common column name = country
# ------------------------------------------------------------

gdp.rename(
    columns={"Country": "country"},
    inplace=True
)

literacy.rename(
    columns={"name": "country"},
    inplace=True
)

print("GDP Columns:")
print(gdp.columns)

print("Literacy Columns:")
print(literacy.columns)
```

```python
# ------------------------------------------------------------
# Q7. Convert required columns into numeric data types
# ------------------------------------------------------------

population["population"] = pd.to_numeric(
    population["population"]
)

gdp["GDP"] = pd.to_numeric(
    gdp["GDP"]
)

internet["internet_users"] = pd.to_numeric(
    internet["internet_users"]
)

literacy["literacy_rate"] = pd.to_numeric(
    literacy["literacy_rate"]
)

print("Population Data Types:")
print(population.dtypes)

print("\nGDP Data Types:")
print(gdp.dtypes)

print("\nInternet Data Types:")
print(internet.dtypes)

print("\nLiteracy Data Types:")
print(literacy.dtypes)
```

```python
# ------------------------------------------------------------
# Q8. Merge all four datasets
# Common column = country
# ------------------------------------------------------------

df = population.merge(
    gdp,
    on="country",
    how="inner"
)

df = df.merge(
    internet,
    on="country",
    how="inner"
)

df = df.merge(
    literacy,
    on="country",
    how="inner"
)

print("Final Merged Dataset:")
print(df)

print("\nShape:")
print(df.shape)
```

```python
# ------------------------------------------------------------
# Q9. Create calculated columns
#
# 1. Internet Penetration (%)
# 2. GDP Per Capita
#
# Round values to 2 decimal places
# ------------------------------------------------------------

df["Internet Penetration (%)"] = (
    df["internet_users"]
    / df["population"]
    * 100
).round(2)

df["GDP Per Capita"] = (
    df["GDP"]
    * 1000000
    / df["population"]
).round(2)

print(df)
```

```python
# ============================================================
# PART D - LOAD DATA
# ============================================================


# ------------------------------------------------------------
# Q10. Create SQLite database
# Store data in table: country_statistics
# ------------------------------------------------------------

import sqlite3

connection = sqlite3.connect(
    "country_statistics.db"
)

df.to_sql(
    "country_statistics",
    connection,
    if_exists="replace",
    index=False
)

print("Database created successfully")

print("Table country_statistics created successfully")

print("Data loaded successfully")
```

```python
# ------------------------------------------------------------
# Q11. Verify that data has been loaded successfully
# Retrieve first five records
# ------------------------------------------------------------

query = """
SELECT *
FROM country_statistics
LIMIT 5;
"""

result = pd.read_sql(
    query,
    connection
)

print(result)
```

```python
# ============================================================
# PART E - PANDAS DATA ANALYSIS
# ============================================================

# Q1. Top 5 Countries by GDP

result = df.nlargest(5, "GDP")
print("\nQ1. Top 5 Countries by GDP")
print(result[["country", "GDP"]])


# Q2. Top 5 Countries by Population

result = df.nlargest(5, "population")
print("\nQ2. Top 5 Countries by Population")
print(result[["country", "population"]])


# Q3. Average Literacy Rate

average_literacy = df["literacy_rate"].mean()

print("\nQ3. Average Literacy Rate")
print(round(average_literacy, 2))


# Q4. Countries with Literacy Rate > 90%

result = df[df["literacy_rate"] > 90]

print("\nQ4. Countries with Literacy Rate > 90%")
print(result[["country", "literacy_rate"]])


# Q5. Countries with Internet Penetration > 70%

result = df[df["Internet Penetration (%)"] > 70]

print("\nQ5. Countries with Internet Penetration > 70%")
print(result[["country", "Internet Penetration (%)"]])


# Q6. Top 5 Countries by GDP Per Capita

result = df.nlargest(5, "GDP Per Capita")

print("\nQ6. Top 5 Countries by GDP Per Capita")
print(result[["country", "GDP Per Capita"]])


# Q7. Total Population

total_population = df["population"].sum()

print("\nQ7. Total Population")
print(total_population)


# Q8. Average Internet Penetration

average_internet = df["Internet Penetration (%)"].mean()

print("\nQ8. Average Internet Penetration")
print(round(average_internet, 2))


# Q9. Country with Highest Internet Users

result = df.loc[df["internet_users"].idxmax()]

print("\nQ9. Country with Highest Internet Users")
print("Country:", result["country"])
print("Internet Users:", result["internet_users"])


# Q10. Correlation between Literacy Rate
#      and Internet Penetration

correlation = df["literacy_rate"].corr(
    df["Internet Penetration (%)"]
)

print("\nQ10. Correlation between Literacy Rate and Internet Penetration")
print(round(correlation, 2))
```

```python
# ============================================================
# PART F - SQL ANALYSIS
# ============================================================

import pandas as pd
import sqlite3

# Connect to SQLite database
connection = sqlite3.connect("country_statistics.db")


# ------------------------------------------------------------
# Q11. Top 5 Countries by GDP
# ------------------------------------------------------------

query = """
SELECT country, GDP
FROM country_statistics
ORDER BY GDP DESC
LIMIT 5;
"""

result = pd.read_sql(query, connection)

print("\nQ11. Top 5 Countries by GDP")
print(result)


# ------------------------------------------------------------
# Q12. Average Literacy Rate
# ------------------------------------------------------------

query = """
SELECT AVG(literacy_rate) AS average_literacy_rate
FROM country_statistics;
"""

result = pd.read_sql(query, connection)

print("\nQ12. Average Literacy Rate")
print(result)


# ------------------------------------------------------------
# Q13. Countries with Literacy Rate > 90%
# ------------------------------------------------------------

query = """
SELECT country, literacy_rate
FROM country_statistics
WHERE literacy_rate > 90;
"""

result = pd.read_sql(query, connection)

print("\nQ13. Countries with Literacy Rate > 90%")
print(result)


# ------------------------------------------------------------
# Q14. Top 5 Countries by Internet Penetration
# ------------------------------------------------------------

query = """
SELECT country, `Internet Penetration (%)`
FROM country_statistics
ORDER BY `Internet Penetration (%)` DESC
LIMIT 5;
"""

result = pd.read_sql(query, connection)

print("\nQ14. Top 5 Countries by Internet Penetration")
print(result)


# ------------------------------------------------------------
# Q15. Total Population
# ------------------------------------------------------------

query = """
SELECT SUM(population) AS total_population
FROM country_statistics;
"""

result = pd.read_sql(query, connection)

print("\nQ15. Total Population")
print(result)


# ------------------------------------------------------------
# Q16. Country with Highest GDP Per Capita
# ------------------------------------------------------------

query = """
SELECT country, `GDP Per Capita`
FROM country_statistics
ORDER BY `GDP Per Capita` DESC
LIMIT 1;
"""

result = pd.read_sql(query, connection)

print("\nQ16. Country with Highest GDP Per Capita")
print(result)


# ------------------------------------------------------------
# Q17. Export Final Merged Dataset as CSV
# ------------------------------------------------------------

df.to_csv("final_merged_dataset.csv", index=False)

print("\nQ17. Final merged dataset exported successfully")


# ------------------------------------------------------------
# Q18. Top 10 Countries by GDP Per Capita
# ------------------------------------------------------------

query = """
SELECT country, `GDP Per Capita`
FROM country_statistics
ORDER BY `GDP Per Capita` DESC
LIMIT 10;
"""

result = pd.read_sql(query, connection)

print("\nQ18. Top 10 Countries by GDP Per Capita")
print(result)


# Close database connection
connection.close()
```
